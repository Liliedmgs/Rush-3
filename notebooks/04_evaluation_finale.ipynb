```python
import pandas as pd

from rush3.core.economie import benefice_pour_1000
from rush3.core.contrats import lire_contrat
from rush3.prediction.evaluation_finale import evaluer_sur_test

resultat = evaluer_sur_test()
resultat
```

```python
validation = {s["nom"]: s["precision_moyenne"] for s in lire_contrat("03_comparaison_modeles")["strategies"]}

comparaison = pd.DataFrame({
    "validation croisée (train)": [validation[resultat["modele"]], validation["Règle simple (voyage personnel)"]],
    "test final": [resultat["precision_modele"], resultat["precision_regle_simple"]],
}, index=[resultat["modele"], "Règle simple (voyage personnel)"])
comparaison["bénéfice test pour 1000 (€)"] = benefice_pour_1000(comparaison["test final"]).round(0)
comparaison
```
